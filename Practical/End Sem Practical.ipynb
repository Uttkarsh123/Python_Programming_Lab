### 1. Scan files to detect:
#### Credit card numbers
#### Aadhaar-like patterns
#### Log file name and line number

```python
import re
file_name = "Records.txt"
#Aadhar pattern 12 digits
credit_pattern = re.compile(r"\b\d{16}\b")
name = input("Enter name: ")
credit = input("Enter Credit Card Number: ")
file = open(file_name, "r")
line_number = 0
found = False
#Searching in file
for line in file:
    line_number += 1
    if name.lower() in line.lower():
        match = credit_pattern.search(line)
        if match and match.group() == credit:
            print(f"Found in {file_name} at line number {line_number}")
            found = True
            break
file.close()
if not found:
    print("Record Not found")
```

### 2. Write a program to calculate Fibonacci series without using recursion

```python
n = int(input("Enter number of terms: "))
a = 0
b = 1
if n >= 1:
    print(a, end=" ")
if n >= 2:
    print(b, end=" ")
#Starting loop from 2 as initial number (0,1) are printed before
for i in range(2, n):
    c = a + b
    print(c, end=" ")
    a = b
    b = c
```

### 3.  Prints n × n stars forming concentric square layers
#### Alternate layers must be filled with stars and spaces
#### Use only nested loops and conditionals
#### Do not use string multiplication or slicing

```python
n = int(input("Enter odd n (>=7): "))
mid = n // 2
for i in range(n):
    for j in range(n):
        #Checking Boundary
        layer = min(i, j, n - i - 1, n - j - 1)
        if layer % 2 == 0:
            print("* ", end="")
        else:
            print("  ", end="")
    print()
```
